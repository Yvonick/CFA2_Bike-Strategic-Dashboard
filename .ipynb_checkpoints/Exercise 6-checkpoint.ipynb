```python
import pandas as pd
import numpy as np
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import matplotlib.pyplot as plt
from datetime import datetime as dt 
import plotly.io as pio
import plotly.express as px

pio.renderers.default = "iframe"
```

```python
# Top 20 bike stations in NYC in 2022 (bar chart)

# Import the trip count by stations we created in the previous exercise

df_tripcount = pd.read_csv(
    "Output/Trip count by stations - NYC 2022.csv",
    engine="pyarrow",                 # faster & lower memory
    dtype_backend="pyarrow",          # compact Arrow dtypes for ints/strings
)

# Calculate the total for each starting station 

df_stations = (
    df_tripcount[['start_station_name', 'trips']]
      .groupby('start_station_name', as_index=False)
      .sum()
)

# Isolate the top20 and export as CSV for re-use in the Streamlit App

df_top20 = df_stations.nlargest(20, 'trips')
df_top20.to_csv('Output/Top20 Bike Stations - NYC 2022.csv')
df_top20
```

```python
# Plot the top20 in a bar chart

# Make a nice color scale that stays clear from the whites

green_scale = px.colors.sequential.Greens[3:]


# Create the figure object

fig = go.Figure(go.Bar(
    x = df_top20['start_station_name'], 
    y = df_top20['trips'],
    marker=dict(
        color=df_top20['trips'],
        colorscale=green_scale
    )
))


# Add title and clean labels

fig.update_layout(
     title = 'Top 20 most popular bike stations in NYC in 2022',
     xaxis_title = 'Start stations',
     yaxis_title ='Sum of trips',
     width = 900, height = 600)

fig.show()
```

```python
# Bike Rides VS Weather NYC 2022 (twin plot)

# Import the CSV file create in an earlier exercise

df_weather = pd.read_csv(
    "Output/Ride Count VS Weather - NYC 2022.csv",
    engine="pyarrow",                 # faster & lower memory
    dtype_backend="pyarrow",          # compact Arrow dtypes for ints/strings
)


# Create the figure object that will host the 2 plots

fig = make_subplots(specs = [[{"secondary_y": True}]])

# Add the first subplot: temperature

fig.add_trace(
    go.Scatter(
        x = df_weather['date'], 
        y = df_weather['avgTemp'], 
        name = 'Daily Average Temperature', 
        marker={'color': df_weather['avgTemp'],'color': 'red'}),
    secondary_y = False)

# Add the second subplot: daily rides

fig.add_trace(
    go.Scatter(
        x = df_weather['date'], 
        y = df_weather['rides'], 
        name = 'Daily bike rides', 
        marker={'color': df_weather['rides'],'color': 'blue'}),
    secondary_y = True)

# Add title

fig.update_layout(
     title = 'Daily Bike Rides VS Daily Average Temperatures - NYC 2022',
     width = 900, height = 600)

fig.show()
```
