```python
import pandas as pd
import numpy as np
```

```python
# Random sampling of the dataset

# Importing the Bike dataframe
# Using pyarrow to read the CSV to save on RAM

df_full = pd.read_csv(
    "Output/bike_data_cleaned_2022.csv",
    engine="pyarrow",                 # faster & lower memory
    dtype_backend="pyarrow",          # compact Arrow dtypes for ints/strings
    parse_dates=["started_at"],       # avoids huge object strings
)

df_full.columns
```

```python
# Sample the dataset

# Drop unnecessary columns
df_trimmed = df_full.drop(columns = {'ride_id', 'start_station_id', 'end_station_id', 'start_lat', 'start_lng', 'end_lat', 'end_lng'})

# Sampling the trimmed dataframe
# Due to the very large dataset, we have to sample 0.5% to get under 25Mb
df_sample = df_trimmed.sample(frac=0.005, random_state=42, ignore_index=True)

df_sample.to_csv('Bike Data - Reduced.csv',index = False)
```
